## Forecast Summary & Interpretation

### Account Ownership (Access)
- **Baseline 2027 projection:** ~35-40% account ownership
- **Range:** pessimistic (24-28%), optimistic (46-52%)
- **Key drivers:** telecom expansion, agent network growth, digital ID rollout

### Digital Payment Adoption (Usage)
- **Baseline 2027 projection:** ~10-15% digital payment penetration
- **Range:** pessimistic (7-11%), optimistic (13-20%)
- **Key drivers:** Telebirr adoption, M-Pesa expansion, merchant enablement, regulatory support

### Scenario Rationale
- **Pessimistic:** Slower regulatory reform, limited merchant adoption, macro headwinds (inflation, currency pressure)
- **Base:** Continued gradual growth, stable policy, moderate investment
- **Optimistic:** Accelerated telecom/mobile money ecosystem, aggressive fintech support, digital ID integration
